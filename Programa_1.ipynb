```python
tabla=int(input("que tabla de multiplicacion quieres"))
limiteinicial=int(input("dame limite inicial"))
limitefinal=int(input("dame limite final"))
while limitefinal <= limiteinicial:
  print("el limite final debe ser mayor que el inicial")
  limitefinal=int(input("dame limite final"))
incrementos=int(input("dame el numero de los incrementos"))
for i in range (limiteinicial, limitefinal + 1, incrementos):
  print(tabla, "*", i, "=", tabla * i)
```
